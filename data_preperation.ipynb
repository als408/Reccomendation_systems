```python
import pandas as pd
import numpy as np

np.random.seed(42)
```

```python
plik = pd.read_csv("/Users/wiktor/Downloads/archive/interactions.csv")

plik.head()
```

```python
click_only = plik[plik["event"] == "click"]
sampled = pd.Series(
    click_only["user"].unique()
).sample(frac=0.1, random_state=42)

filtered = click_only[click_only["user"].isin(sampled)]

quant_sampled_time = np.quantile(filtered["timestamp"], 0.8)
```

```python
filtered_items = filtered.loc[filtered["timestamp"]<quant_sampled_time, "item"]

```

```python
item_counts = filtered_items.value_counts()

top_items = item_counts.head(
    int(len(item_counts) * 0.1)
).index
```
